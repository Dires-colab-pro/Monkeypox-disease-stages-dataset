```python
"""
================================================================================
MPOXSEVNET — TRAIN + CLASS-LEVEL PERFORMANCE ANALYSIS (KAGGLE, SINGLE CELL)
Segmentation: fixed_global (T=125)
Output format:
  Class  Precision  Precision 95% CI  Recall  Recall 95% CI  F1  F1 95% CI  Support
================================================================================
"""

import os, glob, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers, utils
from tensorflow.keras.callbacks import (EarlyStopping, ReduceLROnPlateau,
                                        ModelCheckpoint)
from sklearn.model_selection import train_test_split
from sklearn.metrics import (classification_report, confusion_matrix,
                             precision_recall_fscore_support)
from tqdm import tqdm
warnings.filterwarnings('ignore')

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("=" * 80)
print("MPOXSEVNET — TRAIN + CLASS-LEVEL ANALYSIS (KAGGLE)")
print(f"TensorFlow: {tf.__version__}")
print("Segmentation: fixed_global (T=125)")
print("=" * 80)

# ============================================================================
# 1. LOCATE DATASET
# ============================================================================
def find_data_dir(root='/kaggle/input'):
    for dp, dn, _ in os.walk(root):
        for d in dn:
            if d.startswith('1_'):
                return dp
    return None

DATASET_PATH = find_data_dir()
if DATASET_PATH is None:
    for c in ['/kaggle/input/dirework/mpox-datasett/dataset',
              '/kaggle/input/mpox-datasett/dataset',
              '/kaggle/input/mpox-datasett',
              '/kaggle/input/datasets/dirework/mpox-datasett/dataset',
              '/kaggle/input/datasets/dirework/mpox-datasett']:
        if os.path.isdir(c):
            DATASET_PATH = c; break
if DATASET_PATH is None:
    raise FileNotFoundError("Dataset not found under /kaggle/input")

OUT_DIR = '/kaggle/working'
os.makedirs(OUT_DIR, exist_ok=True)
print(f"✅ Dataset: {DATASET_PATH}")

# ============================================================================
# 2. LOAD IMAGES
# ============================================================================
class_names = ['Macules', 'Papules', 'Vesicles', 'Pustules', 'Scubs', 'Normal']
class_mapping = {
    '1_Macules': 0, '2_Papules': 1, '3_Vesicles': 2,
    '4_Pustules': 3, '5_Scubs': 4, '6_Normal': 5,
}
if not os.path.isdir(os.path.join(DATASET_PATH, '5_Scubs')):
    if os.path.isdir(os.path.join(DATASET_PATH, '5_Scabs')):
        class_mapping = {
            '1_Macules': 0, '2_Papules': 1, '3_Vesicles': 2,
            '4_Pustules': 3, '5_Scabs': 4, '6_Normal': 5,
        }

images, labels = [], []
for folder, lbl in class_mapping.items():
    p = os.path.join(DATASET_PATH, folder)
    if not os.path.exists(p): continue
    files = []
    for ext in ['*.jpg','*.jpeg','*.png','*.JPG','*.JPEG','*.PNG']:
        files.extend(glob.glob(os.path.join(p, ext)))
    print(f"  {folder}: {len(files)} images")
    for f in tqdm(files, desc=f"  load {folder}"):
        img = cv2.imread(f)
        if img is None: continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (224, 224))
        images.append(img); labels.append(lbl)

images = np.array(images, dtype=np.float32) / 255.0
labels = np.array(labels, dtype=np.int32)
print(f"✅ Total: {len(images)} images")

# ============================================================================
# 3. SPLIT 70/15/15
# ============================================================================
X_tr_raw, X_tmp, y_tr, y_tmp = train_test_split(
    images, labels, test_size=0.30, stratify=labels, random_state=SEED)
X_va_raw, X_te_raw, y_va, y_te = train_test_split(
    X_tmp, y_tmp, test_size=0.50, stratify=y_tmp, random_state=SEED)
print(f"Train {len(X_tr_raw)} | Val {len(X_va_raw)} | Test {len(X_te_raw)}")

# ============================================================================
# 4. SEGMENTATION — fixed_global (T=125)
# ============================================================================
def segment_fixed_global(img, T=125):
    u8 = (img * 255).astype(np.uint8)
    gray = cv2.cvtColor(u8, cv2.COLOR_RGB2GRAY)
    gray = cv2.medianBlur(gray, 5)
    _, mask = cv2.threshold(gray, T, 255, cv2.THRESH_BINARY)
    k = np.ones((5, 5), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, k)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN,  k)
    return cv2.bitwise_and(u8, u8, mask=mask).astype(np.float32) / 255.0

def apply_seg(X):
    return np.array([segment_fixed_global(im) for im in tqdm(X, desc="  seg")])

X_tr_seg = apply_seg(X_tr_raw)
X_va_seg = apply_seg(X_va_raw)
X_te_seg = apply_seg(X_te_raw)

# ============================================================================
# 5. HELPERS
# ============================================================================
def to_cat(y):
    y = np.asarray(y)
    return utils.to_categorical(y, num_classes=6) if y.ndim == 1 else y

# ============================================================================
# 6. MODEL
# ============================================================================
def build_mpoxsevnet(lr=1e-3, label_smoothing=0.1):
    base = tf.keras.applications.ResNet50(weights='imagenet', include_top=False,
                                          input_shape=(224, 224, 3))
    base.trainable = False
    x = layers.GlobalAveragePooling2D()(base.output)

    x = layers.BatchNormalization()(x)
    x = layers.Dense(512, activation='relu',
                     kernel_regularizer=regularizers.l2(1e-3))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.5)(x)

    x = layers.Dense(256, activation='relu',
                     kernel_regularizer=regularizers.l2(1e-3))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.5)(x)

    x = layers.Dense(128, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.4)(x)

    out = layers.Dense(6, activation='softmax')(x)
    m = models.Model(base.input, out)
    m.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
        loss=tf.keras.losses.CategoricalCrossentropy(
            label_smoothing=label_smoothing),
        metrics=['accuracy'])
    return m


def unfreeze_conv5_deep(model):
    n = 0
    for l in model.layers:
        if 'conv5_block1' in l.name or \
           'conv5_block2' in l.name or \
           'conv5_block3' in l.name:
            l.trainable = True
            n += 1
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
        loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.1),
        metrics=['accuracy'])
    print(f"  unfroze {n} conv5_block layers")
    return model

# ============================================================================
# 7. CALLBACKS
# ============================================================================
CKPT_PATH = os.path.join(OUT_DIR, 'monkeypox_severity_model.h5')

class CosineLR(tf.keras.callbacks.Callback):
    def __init__(self, base_lr, min_lr, total_epochs):
        super().__init__()
        self.base_lr, self.min_lr, self.total = base_lr, min_lr, total_epochs
    def on_epoch_begin(self, epoch, logs=None):
        c = 0.5 * (1 + np.cos(np.pi * epoch / max(1, self.total)))
        lr = self.min_lr + (self.base_lr - self.min_lr) * c
        self.model.optimizer.learning_rate.assign(lr)

class EMACallback(tf.keras.callbacks.Callback):
    def __init__(self, decay=0.999):
        super().__init__()
        self.decay = decay
        self.ema_weights = None
    def on_train_batch_end(self, batch, logs=None):
        if self.ema_weights is None:
            self.ema_weights = [w.numpy().copy()
                                for w in self.model.trainable_weights]
        else:
            for i, w in enumerate(self.model.trainable_weights):
                self.ema_weights[i] = (self.decay * self.ema_weights[i]
                                       + (1 - self.decay) * w.numpy())

# ============================================================================
# 8. TRAINING
# ============================================================================
EPOCHS_FROZEN = 120
EPOCHS_FT     = 60
BATCH_SIZE    = 32

print("\n" + "=" * 80)
print("Phase 1: frozen base (60 epochs)")
print("=" * 80)

model = build_mpoxsevnet(lr=1e-3, label_smoothing=0.1)

gen_p1 = tf.keras.preprocessing.image.ImageDataGenerator(
    rotation_range=15, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=0.08, zoom_range=0.08, horizontal_flip=True,
    fill_mode='nearest')

cb_p1 = [
    EarlyStopping(monitor='val_accuracy', patience=30,
                  restore_best_weights=True, verbose=0, mode='max'),
    ReduceLROnPlateau(monitor='val_accuracy', factor=0.5,
                      patience=12, min_lr=1e-6, verbose=0, mode='max'),
    ModelCheckpoint(CKPT_PATH, monitor='val_accuracy',
                    save_best_only=True, verbose=1),
]

history1 = model.fit(
    gen_p1.flow(X_tr_seg, to_cat(y_tr), batch_size=BATCH_SIZE),
    validation_data=(X_va_seg, to_cat(y_va)),
    epochs=EPOCHS_FROZEN, callbacks=cb_p1, verbose=1)

print("\n" + "=" * 80)
print("Phase 2: deep fine-tune conv5_block1–3 (30 epochs)")
print("=" * 80)

model = unfreeze_conv5_deep(model)

gen_p2 = tf.keras.preprocessing.image.ImageDataGenerator(
    rotation_range=10, width_shift_range=0.05, height_shift_range=0.05,
    zoom_range=0.05, horizontal_flip=True, fill_mode='nearest')

cb_p2 = [
    CosineLR(base_lr=1e-5, min_lr=1e-7, total_epochs=EPOCHS_FT),
    EarlyStopping(monitor='val_accuracy', patience=24,
                  restore_best_weights=True, verbose=0, mode='max'),
    ModelCheckpoint(CKPT_PATH, monitor='val_accuracy',
                    save_best_only=True, verbose=1),
    EMACallback(decay=0.999),
]

history2 = model.fit(
    gen_p2.flow(X_tr_seg, to_cat(y_tr), batch_size=BATCH_SIZE),
    validation_data=(X_va_seg, to_cat(y_va)),
    epochs=EPOCHS_FT, callbacks=cb_p2, verbose=1)

# Combine histories
history = {
    'accuracy':     history1.history['accuracy']     + history2.history['accuracy'],
    'val_accuracy': history1.history['val_accuracy'] + history2.history['val_accuracy'],
    'loss':         history1.history['loss']         + history2.history['loss'],
    'val_loss':     history1.history['val_loss']     + history2.history['val_loss'],
}

pd.DataFrame({
    'Epoch': range(1, len(history['accuracy']) + 1),
    'Train_Accuracy': history['accuracy'],
    'Val_Accuracy':   history['val_accuracy'],
    'Train_Loss':     history['loss'],
    'Val_Loss':       history['val_loss'],
}).to_csv(os.path.join(OUT_DIR, 'training_history.csv'), index=False)

print("\n✅ Training complete")

# ============================================================================
# 9. PREDICT ON TEST SET
# ============================================================================
y_proba = model.predict(X_te_seg, verbose=0)
y_pred  = np.argmax(y_proba, axis=1)
overall_acc = (y_pred == y_te).mean()
print(f"\nOverall test accuracy: {overall_acc*100:.2f}%")

# ============================================================================
# 10. CLASS-LEVEL METRICS
# ============================================================================
prec, rec, f1, sup = precision_recall_fscore_support(
    y_te, y_pred, labels=list(range(6)), average=None, zero_division=0)

# ============================================================================
# 11. WILSON 95% CI FOR PRECISION AND RECALL
# ============================================================================
def wilson_ci(k, n, z=1.96):
    if n == 0: return (np.nan, np.nan)
    p = k / n
    denom = 1 + z**2 / n
    centre = (p + z**2 / (2 * n)) / denom
    half = z * np.sqrt(p * (1 - p) / n + z**2 / (4 * n**2)) / denom
    return (max(0.0, centre - half), min(1.0, centre + half))

cm = confusion_matrix(y_te, y_pred, labels=list(range(6)))

prec_lo_arr, prec_hi_arr = [], []
rec_lo_arr,  rec_hi_arr  = [], []
for i in range(6):
    TP = cm[i, i]
    FP = cm[:, i].sum() - TP
    FN = cm[i, :].sum() - TP
    pl, ph = wilson_ci(TP, TP + FP)
    rl, rh = wilson_ci(TP, TP + FN)
    prec_lo_arr.append(pl); prec_hi_arr.append(ph)
    rec_lo_arr.append(rl);  rec_hi_arr.append(rh)

# ============================================================================
# 12. BOOTSTRAP 95% CI FOR F1-SCORE
# ============================================================================
def bootstrap_f1_ci(y_true, y_pred, class_idx, n_boot=1000, seed=42):
    rng = np.random.default_rng(seed)
    n = len(y_true)
    scores = []
    for _ in range(n_boot):
        idx = rng.integers(0, n, n)
        yt = y_true[idx]; yp = y_pred[idx]
        TP = np.sum((yt == class_idx) & (yp == class_idx))
        FP = np.sum((yt != class_idx) & (yp == class_idx))
        FN = np.sum((yt == class_idx) & (yp != class_idx))
        if TP == 0 and (FP == 0 or FN == 0):
            scores.append(0.0); continue
        pb = TP / (TP + FP) if (TP + FP) > 0 else 0
        rb = TP / (TP + FN) if (TP + FN) > 0 else 0
        fb = 2 * pb * rb / (pb + rb) if (pb + rb) > 0 else 0
        scores.append(fb)
    scores = np.array(scores)
    return np.percentile(scores, 2.5), np.percentile(scores, 97.5)

f1_lo_arr, f1_hi_arr = [], []
for i in range(6):
    lo, hi = bootstrap_f1_ci(y_te, y_pred, i, n_boot=1000, seed=42)
    f1_lo_arr.append(lo); f1_hi_arr.append(hi)

# ============================================================================
# 13. PRINT TABLE IN EXACT REQUESTED FORMAT
# ============================================================================
print("\n" + "=" * 110)
print("CLASS-LEVEL EVALUATION TABLE")
print("=" * 110)
print(f"{'Class':<10} {'Precision':>9}  {'Precision 95% CI':<18} "
      f"{'Recall':>8}  {'Recall 95% CI':<18} "
      f"{'F1':>6}  {'F1 95% CI':<18} {'Support':>8}")
print("-" * 110)
for i, c in enumerate(class_names):
    prec_ci = f"[{prec_lo_arr[i]:.2f}, {prec_hi_arr[i]:.2f}]"
    rec_ci  = f"[{rec_lo_arr[i]:.2f},  {rec_hi_arr[i]:.2f}]"
    f1_ci   = f"[{f1_lo_arr[i]:.2f},  {f1_hi_arr[i]:.2f}]"
    print(f"{c:<10} {prec[i]:>9.2f}  {prec_ci:<18} "
          f"{rec[i]:>8.2f}  {rec_ci:<18} "
          f"{f1[i]:>6.2f}  {f1_ci:<18} {int(sup[i]):>8d}")
print("=" * 110)

# ============================================================================
# 14. SAVE CSV (same format)
# ============================================================================
full_table = pd.DataFrame({
    'Class': class_names,
    'Precision': [f"{prec[i]:.2f}" for i in range(6)],
    'Precision 95% CI': [f"[{prec_lo_arr[i]:.2f}, {prec_hi_arr[i]:.2f}]"
                         for i in range(6)],
    'Recall': [f"{rec[i]:.2f}" for i in range(6)],
    'Recall 95% CI': [f"[{rec_lo_arr[i]:.2f}, {rec_hi_arr[i]:.2f}]"
                      for i in range(6)],
    'F1': [f"{f1[i]:.2f}" for i in range(6)],
    'F1 95% CI': [f"[{f1_lo_arr[i]:.2f}, {f1_hi_arr[i]:.2f}]"
                  for i in range(6)],
    'Support': [int(sup[i]) for i in range(6)],
})
full_table.to_csv(os.path.join(OUT_DIR, 'class_level_metrics_with_ci.csv'),
                  index=False)
print(f"\n✅ Saved: {OUT_DIR}/class_level_metrics_with_ci.csv")

# ============================================================================
# 15. OVERALL METRICS WITH CI
# ============================================================================
n_total = len(y_te)
n_correct = int((y_pred == y_te).sum())
acc_lo, acc_hi = wilson_ci(n_correct, n_total)

print("\n" + "=" * 80)
print("OVERALL METRICS WITH CI")
print("=" * 80)
print(f"Overall accuracy:  {overall_acc:.3f}  95% CI [{acc_lo:.3f}, {acc_hi:.3f}]")
print(f"Macro precision:   {prec.mean():.3f}")
print(f"Macro recall:      {rec.mean():.3f}")
print(f"Macro F1-score:    {f1.mean():.3f}")

# ============================================================================
# 16. FIGURE 1 — PER-CLASS METRICS WITH ERROR BARS
# ============================================================================
fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(class_names)); width = 0.25

p_vals = np.array([prec[i] for i in range(6)])
r_vals = np.array([rec[i]  for i in range(6)])
f_vals = np.array([f1[i]   for i in range(6)])

p_err = np.vstack([p_vals - np.array(prec_lo_arr),
                   np.array(prec_hi_arr) - p_vals])
r_err = np.vstack([r_vals - np.array(rec_lo_arr),
                   np.array(rec_hi_arr) - r_vals])
f_err = np.vstack([f_vals - np.array(f1_lo_arr),
                   np.array(f1_hi_arr) - f_vals])

ax.bar(x - width, p_vals, width, label='Precision', color='#3498db',
       edgecolor='black', yerr=p_err, capsize=4)
ax.bar(x,         r_vals, width, label='Recall',    color='#2ecc71',
       edgecolor='black', yerr=r_err, capsize=4)
ax.bar(x + width, f_vals, width, label='F1-Score',  color='#f39c12',
       edgecolor='black', yerr=f_err, capsize=4)

ax.set_xticks(x); ax.set_xticklabels(class_names, rotation=20, fontsize=11)
ax.set_ylabel('Score', fontweight='bold', fontsize=12)
ax.set_title('Class-Level Performance with 95% Confidence Intervals',
             fontweight='bold', fontsize=13)
ax.set_ylim(0.6, 1.05); ax.legend(fontsize=11); ax.grid(alpha=0.3, axis='y')

for bars, vals in [(ax.containers[0], p_vals),
                   (ax.containers[1], r_vals),
                   (ax.containers[2], f_vals)]:
    for b, v in zip(bars, vals):
        ax.text(b.get_x() + b.get_width()/2, v + 0.005,
                f"{v:.2f}", ha='center', va='bottom',
                fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, 'class_level_metrics_with_ci.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print(f"✅ Saved: {OUT_DIR}/class_level_metrics_with_ci.png")

# ============================================================================
# 17. FIGURE 2 — NORMALIZED CONFUSION MATRIX
# ============================================================================
cm_norm = cm.astype('float') / cm.sum(axis=1, keepdims=True)
fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names,
            cbar_kws={'label': 'Normalized frequency'}, ax=ax,
            annot_kws={'size': 12, 'weight': 'bold'})
ax.set_title('Normalized Confusion Matrix (Test Set)',
             fontweight='bold', fontsize=13)
ax.set_xlabel('Predicted', fontweight='bold')
ax.set_ylabel('True', fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, 'confusion_matrix.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print(f"✅ Saved: {OUT_DIR}/confusion_matrix.png")

# ============================================================================
# 18. FIGURE 3 — TRAINING CURVES
# ============================================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(history['accuracy'],     label='Train', color='blue', lw=2)
axes[0].plot(history['val_accuracy'], label='Val',   color='red',  lw=2)
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Accuracy')
axes[0].set_title('Training and Validation Accuracy', fontweight='bold')
axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(history['loss'],     label='Train', color='blue', lw=2)
axes[1].plot(history['val_loss'], label='Val',   color='red',  lw=2)
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Loss')
axes[1].set_title('Training and Validation Loss', fontweight='bold')
axes[1].legend(); axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, 'training_curves.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print(f"✅ Saved: {OUT_DIR}/training_curves.png")

# ============================================================================
# 19. FULL CLASSIFICATION REPORT
# ============================================================================
print("\n" + "=" * 80)
print("FULL CLASSIFICATION REPORT")
print("=" * 80)
print(classification_report(y_te, y_pred, target_names=class_names,
                            digits=3, zero_division=0))

# ============================================================================
# 20. SUMMARY
# ============================================================================
print("\n" + "=" * 80)
print("✅ COMPLETE")
print("=" * 80)
print(f"\nFiles saved to {OUT_DIR}:")
print("  monkeypox_severity_model.h5")
print("  class_level_metrics_with_ci.csv")
print("  class_level_metrics_with_ci.png")
print("  confusion_matrix.png")
print("  training_curves.png")
print("  training_history.csv")
```
